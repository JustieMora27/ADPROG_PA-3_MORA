```python
import pandas as pd
```

### **A. Positional and Label-Based Slicing**

#### a. Display shape and complete list of column names

```python
cars = pd.read_csv('cars.csv')
```

```python
print("Shape of DataFrame:", cars.shape)
print("Columns in DataFrame:", cars.columns.tolist())
```

#### b. Using positional slicing, create cars 6 to 10 containing rows 6 through 10 of the dataset, where the first data row is row 1.

```python
cars_6_to_10 = cars.iloc[5:10]
display(cars_6_to_10)
```

#### c. From cars 6 to 10, display only the columns Model, mpg, cyl, hp, and gear, in that order.

```python
selected_columns = cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]
display(selected_columns)
```

### **B. Model Lookup**

#### a. Display the complete row for Toyota Corolla.

```python
toyota = cars[cars['Model'] == 'Toyota Corolla']
display(toyota)
```

#### b. For Pontiac Firebird, display only Model, mpg, hp, and wt.

```python
pontiac = cars[cars['Model'] == 'Pontiac Firebird'][['Model', 'mpg', 'hp', 'wt']]
display(pontiac)
```

### **C. Multi-Model Subsetting**

```python
selected_cars = cars[(cars['Model'] == 'Datsun 710') |
                     (cars['Model'] == 'Lotus Europa') |
                     (cars['Model'] == 'Ferrari Dino')][['Model', 'mpg', 'cyl', 'hp', 'gear']]
display(selected_cars)
```

```python
print("Shape of selected_cars:", selected_cars.shape)
```
